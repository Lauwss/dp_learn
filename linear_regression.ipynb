线性回归(linear regression):

对于一系列解释变量$X_k=[x_1,x_2,...,x_n]^T$，线性回归模型假设因变量$\hat{y}_k$与解释变量之间存在线性关系，即：

$$
\hat{y}_k =\sum_{k=1}^{n}\omega_k x_k+b+\epsilon
$$
其中,$\omega_k$为变量权重.

则预测误差Er可表示为预测值$\hat{y}_k$与真实值$y_k$的差值
$$
Er_k=\lvert y_k-\hat{y}_k\rvert
$$

为评估预估质量,我们构建了平方损失函数(Squared Loss Function):
$$
SLF(\omega,b)_k=\frac{1}{2}Er_{k}^2
$$

则训练损失函数为:
$$
L(\omega,b)=\frac{1}{n}\sum_{k=1}^{n}SLF(\omega,b)_k
$$

通过最小化训练损失函数,我们可以得到最优的参数$\omega,b$:
$$
\omega^*,b^*=\text{arg}\min{_{\omega,b}L(\omega,b)}
$$

由于平方函数是凸函数,训练损失函数为平方函数的求和,故训练损失函数也为凸函数,因此,当训练函数有最小值时,该点的导数值为0,即:
$$
\frac{\partial L(\omega,b)}{\partial \omega}=0,\frac{\partial L(\omega,b)}{\partial b}=0
$$
利用梯度下降法求解即可.

从零开始搭建线性回归模型

```python
import random
import torch
import numpy as np
from d2l import torch as d2l
```

```python
def genertated_data(w, b, num_examples):  #@save
    """生成y = Xw + b + 噪声。"""
    X = torch.normal(0, 1, (num_examples, len(w)))
    y = torch.matmul(X, w) + b
    y += torch.normal(0, 0.01, y.shape)
    return X, y.reshape((-1, 1))

real_w =torch.tensor([2, -3.4, 3.6])
real_b =4.2

features, labels = genertated_data(real_w, real_b, 100000)
```

```python

d2l.set_figsize()
# Generate synthetic data

d2l.plt.scatter(features[:, (0)].detach().numpy(), labels.detach().numpy(), 1);
d2l.plt.show()
```

.detach 函数的作用:将张量从计算图中分离，防止梯度计算

```python
def data_iter(batch_size, features, labels):
    "数据加载器"
    num_examples = len(features)
    indices = list(range(num_examples))
    random.shuffle(indices)  # 样本的读取顺序是随机的
    for i in range(0, num_examples, batch_size):
        batch_indices = torch.tensor(
            indices[i: min(i + batch_size, num_examples)])
        yield features[batch_indices], labels[batch_indices]
        
```

```python
# 初始化参数
w=torch.normal(0, 0.01, size=(3, 1), requires_grad=True)
b=torch.zeros(1, requires_grad=True)
w
```

```python
def linear_regression(X, y):
    "线性回归模型"
    return torch.matmul(X, w) + b
```

```python
def slfmse(y_hat, y):
    "均方损失函数"
    return ((y_hat - y.reshape(y_hat.shape))**2) / 2
```

```python
def sgrad(parameters, lr):
    "小批量随机梯度下降"
    with torch.no_grad():
        for param in parameters:
            param -= lr * param.grad
            param.grad.zero_()
```

```python
lr=0.1
num_epochs=10
batch_size=5
net=linear_regression
loss=slfmse
from tqdm import tqdm
for epoch in tqdm(range(num_epochs),desc='训练中'):
    for X, y in data_iter(batch_size, features, labels):
        l = loss(net(X, y), y)  # 计算小批量损失
        l.sum().backward()  # 小批量的损失对模型参数求梯度
        sgrad([w, b], lr)  # 使用小批量随机梯度下降迭代模型参数
    with torch.no_grad():
        train_l = loss(net(features, labels), labels)
        print(f'epoch {epoch + 1}, loss {float(train_l.mean()):f}')
        
print('训练已完成,参数估计为:', w.reshape(-1), b)
print('误差为:', real_w - w.reshape(-1), real_b - b)
```

利用深度学习框架搭建线性回归模型

```python
import numpy as np
import torch
from torch.utils import data as Data
from d2l import torch as d2l

real_w =torch.tensor([2, -3.4, 3.6])
real_b =4.2

features, labels = d2l.synthetic_data(real_w, real_b, 100000)
```

```python
def load_array(data_arrays, batch_size, is_train=True):
    """构造一个PyTorch数据迭代器。"""
    dataset = Data.TensorDataset(*data_arrays)
    return Data.DataLoader(dataset, batch_size, shuffle=is_train)
```

```python
from torch import nn

net = nn.Sequential(nn.Linear(3, 1))
net[0].weight.data.normal_(0, 0.01)
net[0].bias.data.fill_(0)
loss=nn.MSELoss()
trainer = torch.optim.SGD(net.parameters(), lr=0.1)
num_epochs=10
batch_size=5

for epoch in tqdm(range(num_epochs)):
    for X, y in load_array((features, labels), batch_size):
        l = loss(net(X), y)
        trainer.zero_grad()
        l.backward()
        trainer.step()
    with torch.no_grad():
        train_l = loss(net(features), labels)
        print(f'epoch {epoch + 1}, loss {float(train_l.mean()):f}')

print('训练已完成,参数估计为:', net[0].weight.data.reshape(-1), net[0].bias.data)
```
